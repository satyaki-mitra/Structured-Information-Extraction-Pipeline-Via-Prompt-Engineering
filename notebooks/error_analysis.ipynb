# Error Analysis

Deep dive into model failures and edge cases.

```python
import sys
sys.path.append('..')
import json
from evaluation.error_analysis import ErrorAnalyzer
import pandas as pd
```

```python
with open('../data/annotations/ground_truth_500.json', 'r') as f:
    ground_truth = json.load(f)

# Load predictions (assuming you've run evaluation.ipynb)
with open('experiment_results/llm_evaluation.json', 'r') as f:
    llm_predictions = json.load(f)
```

## Run Error Analysis

```python
analyzer = ErrorAnalyzer()
analysis = analyzer.full_error_analysis(llm_predictions, ground_truth)

print('False Positives:', analysis['false_positives']['count'])
print('False Negatives:', analysis['false_negatives']['count'])
print('Extraction Errors:', analysis['extraction_errors'])
```

## Difficult Cases

```python
difficult = analyzer.identify_difficult_cases(llm_predictions, ground_truth, top_n=10)

for idx, case in enumerate(difficult, 1):
    print(f'\n{idx}. Post ID: {case["post_id"]}')
    print(f'   Errors: {case["error_types"]}')
    print(f'   Description: {case["description"][:100]}...')
```
